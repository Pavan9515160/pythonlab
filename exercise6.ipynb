```python
def palindrome(l):
    n=str(l)
    if n==n[::-1]:
        print("palindrome")
    else:
        print( "not")
(palindrome(121))
palindrome(1452)
palindrome(1221)
```

```python
def rev(l):
  word=""
  for i in range(len(l),0,-1):
    word=word+l[i-1]
  print(word)
rev("pavan")
```

```python
def multiple_values(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg

sum,count,avg=multiple_values(1,2,3,4,5,5,6)
print(f"sum is{sum},length is {count},average is {avg}")
```
