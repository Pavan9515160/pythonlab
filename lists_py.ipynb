```python
#indexing of list
l=[1,2,3,4,5,6,7,8]
m=["pavan",'p','k']
print(l[5])
print(l[6])
print(m[0])

```

```python
#slicing of list
l=[1,2,3,4,5,6,7,8]
print(l[1])
print(l[:2])
print(l[1:4])
print(l[1:8:2])
print(sum(l))
print(l[::-1])#reverse of list
```

```python
#concatenation of list
a=[1,2,3]
b=[4,5,6]
print(a,b)
print(a+b)
```

```python
#repetation of list
a=[1,2,3]
b=[4,5,6]
print(a*3)
print((a+b)*2)
```

```python
#checking
l=[1,2,3,4]
print(100 in l)
print(1 in l)
print(5 not in l)
```

```python
#iteration in list
l=[1,2,3]
for i in l:
  print(i,end=" ")
```

```python
#append and extend
l=[1,2]
l.append(5)
print(l)
l.extend([7,1])
print(l)
```

```python
#list methods
#1.apppend
#2.insert
#3.count
#4.pop
#5.remove
#6.sort
#7.index
#8.extend
l=[1,2,3]
m=[1,2,3,4,1,2,3,1,2,3,1,2,3,31,1,2,3,]
l.pop(1)#pop
print(l)
l.remove(3)#remove
print(l)
l.insert(1,5)#insert
print(m.count(1))#count
print(l)
print(m.sort())#sort
print(l.index(1))#index

```
