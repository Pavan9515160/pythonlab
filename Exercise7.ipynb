```python
#set operations
a={1,2,3,4,5}
b={3,4,5}
a.issubset(b)
print(a|b) #union
print(a&b) #intersection
print(a-b) #difference
print(a^b) #symmetric difference
```

```python
#set methods
s={1,2,3,4,5}
t={2,3,4,5,6,7,8}
s.add(7) #add
print(s)
s.remove(7) #remove
print(s)
s.discard(8) #discard
print(s)
s.pop() #pop
print(s)
s.clear() #pop
print(s)
t.issuperset(s)
```

```python
s.isdisjoint(t)
```
