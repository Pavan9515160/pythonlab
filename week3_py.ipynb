```python
#print even numbers
n=int(input("enter number"))
for i in range(0,n,2):
  print(i,end=" ")
```

```python
#check given number is prime or not
n=int(input("enter number"))
if n==2 or n==3 or n==5 :
  print("prime")
elif n%2==0 or n%3==0 or n%5==0 :
  print("not prime")
else :
  print("prime")
```

```python
#print prime numbers between two given numbers
n1=int(input("enter number"))
n2=int(input("enter number"))
for i in range(n1,n2+1):
    if i==2 or i==3 or i==5 :
      print(i,end=" ")
    elif i%2==0 or i%3==0 or i%5==0 :
      continue
    elif i==1:
      continue
    else :
      print(i,end=" ")
```

```python
#print the sum of digits in given number
number=int(input("enter number"))
total = 0

while number > 0:
    digit = number % 10
    total += digit
    number //= 10

print("Sum of digits:", total)
```

```python
#print 1  to n numbers in 10 times
n=int(input("enter number"))
for i in range(0,11):
  print(end="\n")
  for i in range(0,n+1):
    print(i,end=" ")
```
